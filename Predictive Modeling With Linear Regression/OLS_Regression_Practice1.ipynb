```python
X = df['Credit amount']
y = df['Age']
```

Step 4 : Add constants to features

```python
X = sm.add_constant(X)
```

(Step 5: Fit the OLS Model
Create and fit the OLS model using X and Y and store the result in model.

```python
model = sm.OLS(y,X).fit()
print(model)
```

Step 6. Retrieve MSE Total
Get the mean squared error total and assign it to mse_total

```python
mse_total = model.mse_total
print(mse_total)
```
