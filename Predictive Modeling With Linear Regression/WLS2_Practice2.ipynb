```python
# Step 1: Import libraries
import pandas as pd
from sklearn.model_selection import train_test_split
import statsmodels.api as sm
from statsmodels.stats.outliers_influence import variance_inflation_factor

# Step 2: Load dataset
df = pd.read_csv('data.csv')

# Step 3: Prepare features and target
X = df[['mpg', 'cylinders', 'displacement', 'horsepower', 'weight', 'acceleration', 'model year', 'origin']]
y = X['mpg']
X.drop('mpg', axis=1, inplace=True)

# Step 4: Encode categorical variable
X_dummies = pd.get_dummies(X, columns=['model year'], drop_first=True, dtype=float)

# Step 5: Train-test split
X_train, X_test, y_train, y_test = train_test_split(X_dummies, y, test_size=0.3, random_state=1)

# Step 6: Fit OLS model
X_train = sm.add_constant(X_train)
ols = sm.OLS(y_train, X_train).fit()
print(ols.summary())

# Step 7: Calculate VIFs (excluding 'const')
vif = [variance_inflation_factor(X_train.values, i) for i in range(1, X_train.shape[1])]
vif_dict = dict(zip(X_train.columns[1:], vif))
print("VIF values:", vif_dict)

# Step 8: Identify predictor with maximum VIF
max_vif = max(vif_dict, key=vif_dict.get)
print("Predictor with maximum VIF:", max_vif)
```
