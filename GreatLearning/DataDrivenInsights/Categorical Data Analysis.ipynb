```python
import pandas as pd
```

```python
df=pd.read_csv('/GreatLearning/datafiles/userData.csv')
```

```python
housing_count = df['Housing'].unique()
print(housing_count)
```

```python
housing_count = df['Housing'].value_counts()
print(housing_count)
```

```python
categorical_columns = ['Sex','Job','Housing','Saving accounts','Checking account','Purpose']
unique_data = {
    'Column': [],
    'Unique Values': []
}
for column in categorical_columns:
    unique_data['Column'].append(column)
    unique_data['Unique Values'].append(df[column].unique())
print(unique_data['Column'])
print(unique_data['Unique Values'])

unique_values_df = pd.DataFrame(unique_data)
print(unique_values_df)
#unique_values_df  = df['Sex','Job','Housing','Saving accounts','Checking account','Purpose'].value_counts()
#print(unique_values_df)
```

```python
mode_value_from_purpose = df['Purpose'].mode()
print(mode_value_from_purpose)

most_common = df['Purpose'].value_counts(normalize=True).idxmax()
print(most_common)
percentage = round(df['Purpose'].value_counts(normalize=True).max() * 100, 2)
print(percentage)
majority = {most_common: percentage}
print(majority)
```
