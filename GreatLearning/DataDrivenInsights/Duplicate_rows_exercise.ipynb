```python
import pandas as pd
```

```python
df = pd.read_csv('/GreatLearning/datafiles/Melbourne_Housing.csv')
df.head()
```

```python
num_duplicates = df.duplicated().sum()
print('number of duplicate rows:', num_duplicates)
```

```python
# Step 3: Remove duplicate rows in-place
df.drop_duplicates(inplace=True)

# Optional: confirm how many rows remain
print("Remaining rows after removing duplicates:", len(df))
```

```python
num_nonduplicates = len(df) - len((df.duplicated()))
print(num_nonduplicates)
```

```python
num_nonduplicates =  (~df.duplicated()).sum()
print('number of non-duplicate rows:', num_nonduplicates)
```

```python
df.drop_duplicates(inplace=True)
df.reset_index(drop=True, inplace=True)
```
