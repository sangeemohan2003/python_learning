```python
import pandas as pd
import numpy as np

# Step 2: Read Dataset
df = pd.read_csv('data.csv')

# Step 3: Compute squared values of 'Price'
sq_price = df['Price'] ** 2

# Step 4: Compute sum of squares
sum_sq_price = sq_price.sum()

# Step 5: Compute norm (L2)
norm = np.sqrt(sum_sq_price)

# Step 6: Normalize 'Price' column
def normalize_price(x):
    return x / norm

df['Price'] = df['Price'].apply(normalize_price)
```
