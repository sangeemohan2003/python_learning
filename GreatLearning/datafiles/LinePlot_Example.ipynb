```python
import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
revenue = [5000, 5200, 5800, 6000, 6300]

plt.plot(months, revenue,marker='o')
plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue ($)")
plt.grid(True)
#plt.colorbar(label="Revenue")
plt.show()
```
