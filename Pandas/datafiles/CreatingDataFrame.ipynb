```python
import pandas as pd
```

```python
data = {
    'name': ['Mohanraj', 'Subramaniam', 'Niranjan', 'Sangeetha', 'Pinky'],
    'age': [47, 47, 18, 42, 42],
    'city': ['Sathyamanagalam','Sathyamangalam','Jersey City','Anamalai','Coimbatore' ]
}
row_labels  = [1,2,3,4,5]
df = pd.DataFrame(data=data, index=row_labels)
print(df)
```

Accessing data from dataframe:   Accessing a column in a pandas data frame the same way you would get a vlue from a dictionary.

```python
age = df['age']
print(age)
```

Access method 2 -  this method will not work if the column header contains spaces or special characters such as asterisk

```python
age = df.age
print(age)
```

it is important to notice that you have extracted both the data and the corresponding row labels. Each column of a pandas Dataframe is an instance of pandas.Series, a structure that holds one-dimensional data along with their row labels.

Accessing a single item of pandas.Series object using their label

```python
age[1]
print(age[1])
cities = df['city']
print(cities)
```
