```python
df = pd.read_csv('/Users/mohanrajsubramaniam/PycharmProjects/python_learning/Pandas/datafiles/userData.csv')
df.head()
corr_matrix = df.corr(numeric_only=True)
print(corr_matrix)
max_corr = corr_matrix.get('Age')
print("max_corr" , max_corr)

plot = sns.heatmap(data=corr_matrix, annot=True)
```

```python
for col in corr_matrix.columns:
    corr_matrix.loc[col, col] = pd.NA

# Flatten and drop NaNs
corr_flat = corr_matrix.stack().dropna()

# Get maximum correlation value
max_corr = corr_flat.max()

print("Maximum correlation (excluding 1.0):", max_corr)
```
