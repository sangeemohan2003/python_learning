```python
import pandas as pd
import seaborn as sns
from matplotlib import pyplot as plt
```

```python
# plot without any bins
df=pd.read_csv('/Users/mohanrajsubramaniam/PycharmProjects/python_learning/Pandas/datafiles/data.csv')
plot = sns.histplot(data=df,x="Age")
plt.show()
```

```python
# Plot the Histogram of Age Column with bins=10 specified
plot=sns.histplot(data=df,x='Age',bins=10)
plt.show()
```

```python
# Plot the histogram of the age column with  binwidth = 5 specified
plot=sns.histplot(data=df,x='Age',binwidth=5)
plt.show()
```

```python
plot=sns.histplot(data=df,x='Price')
plt.show()
```

```python
plot=sns.histplot(data=df,x='Price',bins=5)
plt.show()
```
